## Credit Card fraud detection

```python
import numpy as np
import pandas as pd

df = pd.read_csv('creditcard.csv')
df.head(), df.shape
```

```python
df.describe()
df.Class.value_counts()
```

```python
fraud = df[df['Class'] == 1]
valid = df[df['Class'] == 0]

outlier = len(fraud)/float(len(valid))
print(outlier)
print('Fraud Cases: {}'.format(len(df[df['Class'] == 1])))
print('Valid Transactions: {}'.format(len(df[df['Class'] == 0])))
```

***Comparing the fraud and valid transactions amount***

```python
print("Amount details of fraud transactions")
fraud.Amount.describe()
```

```python
print("Amount details of valid transactions")
valid.Amount.describe()
```

#### Plotting correlation matrix

```python
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib import gridspec

corr_matrix = df.corr()
fig = plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, vmax=0.8, square=True)
plt.show()
```

##### Splitting the dataset

```python
X = df.iloc[:,:-1]
y = df.iloc[:,-1]

X.shape, y.shape
```

```python
# convert the pandas DataFrame to numpy arrays for faster computation
xData, yData = X.values, y.values
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(xData, yData, test_size=0.2, random_state=42)
```

### Train the model (Random Forest Classifier)

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_pred, y_test)}")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, matthews_corrcoef, confusion_matrix

accuracy = accuracy_score(y_pred, y_test)
precision = precision_score(y_pred, y_test)
recall = recall_score(y_pred, y_test)
f1 = f1_score(y_pred, y_test)
mcc = matthews_corrcoef(y_pred, y_test)

print("Model Evaluation Metrics:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
print(f"Matthews Correlation Coefficient: {mcc:.4f}")
```

#### plotting the confusion matrix

```python
conf_matrix = confusion_matrix(y_pred, y_test)

plt.figure(figsize=(8, 6))
sns.heatmap(
  conf_matrix, annot=True, 
  fmt="d", cmap="Blues",
  xticklabels=['Normal', 'Fraud'], 
  yticklabels=['Normal', 'Fraud']
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.show()
```
