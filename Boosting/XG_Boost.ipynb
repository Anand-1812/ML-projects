# Implementing XGBoost

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('Churn_Modelling.csv')
df.head()
```

```python
df.describe()
```

```python
X = df.iloc[:, 3:13]
y = df.iloc[:, 13].values

X.shape, y.shape
```

#### converts the specified columns to the categorical data type :- Because XGBoost can handle categorical features.

```python
X['Geography'] = X['Geography'].astype('category')
X['Gender'] = X['Gender'].astype('category')
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

X_train.shape, y_train.shape
```

#### We will convert our dataset into DMatrix, because DMatrix is a special data structure in XGBoost for faster training and less memory use.

```python
import xgboost as xgb

xgb_train = xgb.DMatrix(X_train, y_train, enable_categorical=True)
xgb_test = xgb.DMatrix(X_test, y_test, enable_categorical=True)
```

**Creating the XGBoost model**

```python
params = {
  'objective': 'binary:logistic',
  'max_depth': 3,
  'learning_rate': 0.1,
}
n=50
model = xgb.train(params=params,dtrain=xgb_train,num_boost_round=n)
```

```python
preds = model.predict(xgb_test)
preds = np.round(preds)
preds
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, preds)
print(f"Accuracy: {accuracy}")
```
